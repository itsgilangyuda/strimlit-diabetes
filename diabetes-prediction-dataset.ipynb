```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df=pd.read_csv("/kaggle/input/datasets/iammustafatz/diabetes-prediction-dataset/diabetes_prediction_dataset.csv")
df
```

```python
from sklearn.preprocessing import LabelEncoder
xy=LabelEncoder()
df["gender"]=xy.fit_transform(df["gender"])
```

```python
df
```

```python
df=pd.get_dummies(df,columns=["smoking_history"])
df
```

```python
x=df.drop("diabetes",axis=1)
y=df["diabetes"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.2,random_state=10)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
lg=LogisticRegression()
```

```python
lg.fit(X_train,Y_train)
```

```python
pred=lg.predict(X_test)
```

```python
print(type(pred))
print(len(pred))
```

```python
pred[:10]
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(Y_test, pred)
print(accuracy)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(Y_test, pred))
print(confusion_matrix(Y_test, pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
```

```python
knn.fit(X_train, Y_train)
```

```python
knn_pred = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(Y_test,knn_pred)
print(accuracy)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(Y_test,knn_pred))
print(confusion_matrix(Y_test,knn_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
tree = DecisionTreeClassifier(random_state=42)
```

```python
tree.fit(X_train, Y_train)
```

```python
tree_pred = tree.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(Y_test,tree_pred)
print(accuracy)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(Y_test,tree_pred))
print(confusion_matrix(Y_test,tree_pred))
```

```python
from sklearn.metrics import accuracy_score
print("Logistic:", accuracy_score(Y_test,pred))
print("KNN:",accuracy_score(Y_test, knn_pred))
print("Decision Tree:",accuracy_score(Y_test, tree_pred))
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_test, pred))
print(classification_report(Y_test, knn_pred))
print(classification_report(Y_test, tree_pred))
```
